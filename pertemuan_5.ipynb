```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt

path = '/content/drive/MyDrive/PCVK_2026/'
filenames = ['ktm1a.jpeg', 'ktm1b.jpeg', 'ktm1c.jpeg', 'ktm1d.jpeg']

# Membuat figure 2x2
fig, axs = plt.subplots(2, 2, figsize=(14, 10))
axs = axs.flatten()  # Mengubah matriks 2x2 menjadi array 1D agar mudah diiterasi

for i, fname in enumerate(filenames):
    # Load citra sebagai grayscale
    img = cv.imread(path + fname, cv.IMREAD_GRAYSCALE)

    if img is None:
        print(f"File {fname} tidak ditemukan di path yang ditentukan.")
        continue

    # 1. Hitung histogram frekuensi absolut menggunakan OpenCV
    hist = cv.calcHist([img], [0], None, [256], [0, 256]).flatten()

    # 2. Hitung Histogram Ternormalisasi p(j) = hist / total_pixel
    total_pixels = img.shape[0] * img.shape[1]
    p_j = hist / total_pixels

    # 3. Hitung Cumulative Distribution Function (CDF)
    cdf = p_j.cumsum()

    # Sumbu x untuk nilai intensitas 0-255
    bins = np.arange(256)

    # Plot Histogram Ternormalisasi p(j) (Sumbu Y kiri)
    color = 'tab:blue'
    axs[i].set_xlabel('Intensitas Warna (j)')
    axs[i].set_ylabel('Probabilitas p(j)', color=color)
    axs[i].bar(bins, p_j, color=color, alpha=0.6, width=1.0, label='Histogram p(j)')
    axs[i].tick_params(axis='y', labelcolor=color)

    # Plot Kurva CDF (Sumbu Y kanan)
    ax2 = axs[i].twinx()  # Buat sumbu Y kedua yang terintegrasi
    color = 'tab:red'
    ax2.set_ylabel('CDF', color=color)
    ax2.plot(bins, cdf, color=color, linewidth=2, label='CDF')
    ax2.tick_params(axis='y', labelcolor=color)
    ax2.set_ylim([0, 1.05])

    axs[i].set_title(f'Histogram & CDF: {fname}')
    axs[i].grid(True, linestyle='--', alpha=0.5)

plt.suptitle('244107020029 - Perbandingan Histogram Ternormalisasi p(j) dan Kurva CDF (KTM1a - KTM1d)', fontsize=16)
plt.tight_layout(rect=[0, 0, 1, 0.96])
plt.show()
```

Penjelasan perbedaan bentuk curva cdf:
1. citr paling gelap, Bentuk Kurva naik secara sangat tajam di bagian awal (area intensitas rendah/sisi kiri grafik) dan langsung mendatar (mencapai nilai 1.0) sebelum mencapai nilai intensitas menengah.

penyebabnya, Sebagian besar piksel pada citra gelap terkonsentrasi pada nilai intensitas yang rendah (0-100). Karena CDF merupakan akumulasi nilai $p(j)$, akumulasi probabilitas berlangsung cepat di area gelap.

2. citra paling terang, Bentuk Kurva berada pada nilai yang tetap rendah (hampir mendatar dekat angka 0) di area awal, kemudian baru melonjak curam ke angka 1.0 di bagian akhir (area intensitas tinggi/sisi kanan grafik).

penyebabnya, piksel bertumpuk pada nilai intensitas tinggi (150–255), sehingga tidak ada akumulasi nilai yang signifikan di nilai intensitas rendah dan sedang. Akumulasi probabilitas baru melonjak ketika mendekati batas intensitas maksimal.

```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt

def process_histogram_equalization_rgb(image_path, title_prefix=""):
    # 1. Read Image
    img = cv.imread(image_path)
    if img is None:
        print(f"Gambar pada path '{image_path}' tidak ditemukan.")
        return

    # Konversi BGR ke RGB
    img_rgb = cv.cvtColor(img, cv.COLOR_BGR2RGB)
    img_equalized = np.zeros_like(img_rgb)

    # Simpan histogram tiap channel
    hist_before = {}
    hist_after = {}
    colors = ('red', 'green', 'blue')

    # 2. Histogram Equalization per channel (R, G, B)
    for c, color in enumerate(colors):
        # Hitung Histogram Sebelum
        hist, _ = np.histogram(img_rgb[:, :, c].flatten(), 256, [0, 256])
        hist_before[color] = hist

        # Hitung Normalized Histogram p(j) & CDF
        p_j = hist / (img_rgb.shape[0] * img_rgb.shape[1])
        cdf = p_j.cumsum()

        # Mapping S(j)
        s_j = np.round(255 * cdf).astype(np.uint8)
        img_equalized[:, :, c] = s_j[img_rgb[:, :, c]]

        # Hitung Histogram Sesudah
        hist_eq, _ = np.histogram(img_equalized[:, :, c].flatten(), 256, [0, 256])
        hist_after[color] = hist_eq

    # Visualisasi Hasil: Citra & Histogram RGB (Sebelum dan Sesudah)
    names = np.arange(256)
    fig, axs = plt.subplots(2, 3, figsize=(15, 6), sharex=True, sharey=True)
    fig.suptitle('244107020029 - Histogram RGB plot', fontsize=14)

    # Baris 1: Histogram Sebelum (Red, Green, Blue)
    axs[0, 0].bar(names, hist_before['red'], color='red', width=1.0)
    axs[0, 1].bar(names, hist_before['green'], color='green', width=1.0)
    axs[0, 2].bar(names, hist_before['blue'], color='blue', width=1.0)

    # Baris 2: Histogram Sesudah (Red, Green, Blue)
    axs[1, 0].bar(names, hist_after['red'], color='red', width=1.0)
    axs[1, 1].bar(names, hist_after['green'], color='green', width=1.0)
    axs[1, 2].bar(names, hist_after['blue'], color='blue', width=1.0)

    # Label Sumbu X dan Y
    fig.text(0.08, 0.5, 'Jumlah Kemunculan', va='center', rotation='vertical', fontsize=11)
    fig.text(0.5, 0.04, 'Intensitas Warna', ha='center', fontsize=11)

    plt.tight_layout(rect=[0.08, 0.05, 1, 0.95])
    plt.show()

    # Opsional: Tampilkan Juga Citra Sebelum dan Sesudah
    fig_img, axs_img = plt.subplots(1, 2, figsize=(10, 5))
    axs_img[0].imshow(img_rgb)
    axs_img[0].set_title(f'Sebelum (Original) - {title_prefix}')
    axs_img[0].axis('off')

    axs_img[1].imshow(img_equalized)
    axs_img[1].set_title(f'Sesudah (Equalized) - {title_prefix}')
    axs_img[1].axis('off')

    plt.tight_layout()
    plt.show()

path_lena = '/content/drive/MyDrive/PCVK_2026/lena_lc.jpg'
process_histogram_equalization_rgb(path_lena, title_prefix="lena_lc.jpg")

path_ktm = '/content/drive/MyDrive/PCVK_2026/ktm1a.jpeg'
process_histogram_equalization_rgb(path_ktm, title_prefix="ktm1a.jpeg")
```

```python
import cv2
import numpy as np

def manual_equalize_hist(channel):
    # Hitung histogram
    hist, _ = np.histogram(channel.flatten(), 256, [0, 256])
    # Hitung PDF / p(j)
    p_j = hist / channel.size
    # Hitung CDF
    cdf = p_j.cumsum()
    # Pemetaan S(j) = round(255 * CDF)
    s_j = np.round(255 * cdf).astype(np.uint8)
    return s_j[channel]

def compare_equalization(image_path):
    image = cv2.imread(image_path)
    if image is None:
        print(f"Gambar pada path '{image_path}' tidak ditemukan.")
        return

    # Split channel (BGR)
    b, g, r = cv2.split(image)

    b_equalized = cv2.equalizeHist(b)
    g_equalized = cv2.equalizeHist(g)
    r_equalized = cv2.equalizeHist(r)
    img_cv2 = cv2.merge([b_equalized, g_equalized, r_equalized])

    b_manual = manual_equalize_hist(b)
    g_manual = manual_equalize_hist(g)
    r_manual = manual_equalize_hist(r)
    img_manual = cv2.merge([b_manual, g_manual, r_manual])

    # 3. Menghitung Selisih Maksimum per Channel
    diff_b = np.max(np.abs(b_equalized.astype(int) - b_manual.astype(int)))
    diff_g = np.max(np.abs(g_equalized.astype(int) - g_manual.astype(int)))
    diff_r = np.max(np.abs(r_equalized.astype(int) - r_manual.astype(int)))
    diff_total = np.max(np.abs(img_cv2.astype(int) - img_manual.astype(int)))

    print(f"Hasil selisih maksimum ({image_path.split('/')[-1]})")
    print(f"Selisih Channel Blue  : {diff_b}")
    print(f"Selisih Channel Green : {diff_g}")
    print(f"Selisih Channel Red   : {diff_r}")
    print(f"Selisih Maksimum Total: {diff_total}\n")

compare_equalization('/content/drive/MyDrive/PCVK_2026/lena_lc.jpg')
compare_equalization('/content/drive/MyDrive/PCVK_2026/ktm1a.jpeg')
```

Penjelasan:
1. pada citra lena_lc.jpg (Selisih = 0), Selisih maksimum bernilai 0 pada semua channel (Blue, Green, Red). Artinya, hasil ekualisasi histogram antara fungsi cv2.equalizeHist dan metode manual pada citra ini 100% identik.

2. pada ktm1a.jpeg (Selisih Tidak Nol), Terdapat perbedaan selisih sebesar 1 pada channel Blue dan Green, serta 6 pada channel Red. Penyebab selisih ini terjadi karena perbedaan detail algoritma di balik layar:

```python
# Cara 2: hanya kanal kecerahan yang diekualisasi
ycrcb = cv.cvtColor(bgr, cv.COLOR_BGR2YCrCb)
y, cr, cb = cv.split(ycrcb)
he_y = cv.cvtColor(cv.merge([cv.equalizeHist(y), cr, cb]), cv.COLOR_YCrCb2BGR)

# varian setara pada ruang warna lain
hsv = cv.cvtColor(bgr, cv.COLOR_BGR2HSV)
h, sat, v = cv.split(hsv)
he_v = cv.cvtColor(cv.merge([h, sat, cv.equalizeHist(v)]), cv.COLOR_HSV2BGR)

lab = cv.cvtColor(bgr, cv.COLOR_BGR2LAB)
l, a, b = cv.split(lab)
he_l = cv.cvtColor(cv.merge([cv.equalizeHist(l), a, b]), cv.COLOR_LAB2BGR)

judul = ['Asli', 'Ekualisasi B, G, R', 'Kanal Y', 'Kanal V', 'Kanal L']
citra = [bgr, he_rgb, he_y, he_v, he_l]

plt.figure(figsize=(18, 4))
for i, (c, t) in enumerate(zip(citra, judul), 1):
    plt.subplot(1, 5, i)
    plt.imshow(cv.cvtColor(c, cv.COLOR_BGR2RGB))
    plt.title(NIM + ' - ' + t, fontsize=9)
    plt.axis('off')
plt.tight_layout()
plt.show()

def geser_hue(asli, hasil):
    h1 = cv.cvtColor(asli, cv.COLOR_BGR2HSV)[:, :, 0].astype(np.int16)
    h2 = cv.cvtColor(hasil, cv.COLOR_BGR2HSV)[:, :, 0].astype(np.int16)
    d = np.abs(h1 - h2)
    d = np.minimum(d, 180 - d)  # Hue melingkar 0-179
    return d.mean()

print('%-22s %16s %16s' % ('Cara', 'geser Hue (der)', 'rerata terang'))
for t, c in zip(judul, citra):
    gh = 0.0 if t == 'Asli' else geser_hue(bgr, c) * 2
    rk = cv.cvtColor(c, cv.COLOR_BGR2GRAY).mean()
    print('%-22s %16.2f %16.1f' % (t, gh, rk))
```
